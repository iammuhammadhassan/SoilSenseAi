
# SoilSenseAI — System Methodology

## 1. Project Objective

SoilSenseAI is a Punjab-wide, coordinate-based soil intelligence system.

Its purpose is to accept a geographical location (latitude and longitude), assess the available soil information for that location, identify potential soil-related risks, and provide evidence-based agricultural recommendations.

The system will initially use two spatial data sources:

1. **APNA Farm:** Digitized Punjab soil maps containing pH, electrical conductivity (EC), phosphorus (P), potassium (K), organic matter (OM), and soil texture.
2. **ISRIC SoilGrids:** Modelled soil-property estimates covering Punjab at approximately 250 m resolution, across multiple soil depths.

## 2. Expected System Workflow

1. User provides latitude and longitude.
2. System checks whether the location falls within the Punjab study area.
3. System retrieves the corresponding APNA Farm and SoilGrids data.
4. System processes and assesses available soil properties.
5. System identifies potential soil limitations and risks.
6. System evaluates soil-based crop suitability.
7. System generates management and fertilizer recommendations where sufficient evidence exists.
8. System presents results with data sources, confidence, and limitations.

## 3. Main Soil Properties

| Property | Primary source | Intended use |
|---|---|---|
| pH | APNA Farm + SoilGrids | Soil acidity/alkalinity assessment |
| EC | APNA Farm | Potential salinity assessment |
| Phosphorus | APNA Farm | Nutrient availability assessment |
| Potassium | APNA Farm | Nutrient availability assessment |
| Organic Matter | APNA Farm | Soil organic matter assessment |
| Soil Texture | APNA Farm + SoilGrids | Soil physical characteristics |
| Clay, Sand, Silt | SoilGrids | Texture composition |
| SOC | SoilGrids | Soil organic carbon assessment |
| Nitrogen | SoilGrids | Soil nitrogen assessment |
| CEC | SoilGrids | Nutrient-retention capacity |
| Bulk Density | SoilGrids | Soil compaction and physical condition |
| Water Content | SoilGrids | Soil water-retention characteristics |

## 4. Important Data Principles

- APNA Farm values represent digitized map classes, not verified laboratory measurements.
- SoilGrids values are modelled estimates, not direct measurements at the user's location.
- APNA Farm and SoilGrids must retain separate source identifiers.
- Soil Organic Carbon (SOC) and Organic Matter (OM) are related but are not interchangeable.
- SoilGrids does not provide equivalent available phosphorus, potassium, or electrical conductivity layers.
- Missing values must remain identifiable; they must not be silently replaced with fabricated values.
- Recommendations must reflect data availability and uncertainty.

## 5. Final Product

The intended product is a coordinate-based soil intelligence application that provides:

- Soil property assessment
- Soil health indicators
- Potential soil risk identification
- Soil-based crop suitability ranking
- Fertilizer and soil-management guidance
- Spatial visualization
- Data-source and confidence information



## 6. Coordinate-Based Data Retrieval

### 6.1 User Input

The system accepts:

- Latitude
- Longitude

Coordinates must be valid geographic coordinates and fall within the Punjab study area.

### 6.2 APNA Farm Data Retrieval

APNA Farm datasets contain digitized map cells represented by geographic coordinates.

For a user-provided location:

1. Search for the nearest available APNA Farm map cell.
2. Calculate the geographic distance between the requested location and that cell.
3. Accept the result only if it falls within an appropriate spatial matching tolerance.
4. Retrieve the available soil-property classes and their associated value intervals.
5. Report missing properties rather than inventing values.

**Important:** APNA Farm values represent digitized map information. The nearest cell is an estimate for the requested location, not a laboratory sample collected at that exact coordinate.

The matching tolerance must be established from the actual grid spacing and spatial validation.

### 6.3 SoilGrids Data Retrieval

SoilGrids provides spatial raster estimates at approximately 250 m resolution.

For a user-provided location:

1. Transform the input coordinates into the raster's coordinate reference system.
2. Identify the raster pixel containing the location.
3. Extract the required soil-property values.
4. Retrieve multiple depth intervals where relevant.
5. Convert raw raster values into their documented measurement units.
6. Preserve missing or NoData values.

SoilGrids estimates must be identified separately from APNA Farm values.

### 6.4 Combining Both Sources

The system will combine the retrieved information into a single location-based assessment.

However:

- Values from different sources must not be treated as identical measurements.
- APNA Farm and SoilGrids pH can be compared as separate estimates.
- APNA Farm organic matter and SoilGrids soil organic carbon must remain separate properties.
- SoilGrids texture composition can complement APNA Farm texture classes.
- EC, phosphorus, and potassium will use APNA Farm information unless additional reliable datasets become available.

### 6.5 Location Assessment Output

For every requested coordinate, the retrieval module should return:

- Input latitude and longitude
- Punjab boundary validation
- APNA Farm values and availability
- SoilGrids values and availability
- SoilGrids depth information
- Distance to matched APNA Farm cell
- Source of each value
- Missing-data indicators

### 6.6 Spatial Accuracy and Limitations

The system must not claim that every coordinate has an exact soil measurement.

APNA Farm's digitized map grid is substantially coarser than SoilGrids' raster resolution. Their spatial coverage and uncertainty therefore differ.

The application must communicate these differences to users and avoid presenting estimated values as verified field measurements.



## 7. Soil Assessment Methodology

### 7.1 Objective

The soil assessment module converts retrieved soil properties into understandable indicators of soil condition.

It will evaluate individual soil properties, identify potential limitations, and combine the available evidence into a location-based assessment.

The module must distinguish between:
- Observed or laboratory-measured values, when available.
- APNA Farm digitized map estimates.
- SoilGrids modelled estimates.

### 7.2 Property-Based Assessment

Each soil property will be evaluated independently.

| Property | Assessment | Potential concern |
|---|---|---|
| pH | Acidity or alkalinity | Nutrient availability limitations |
| EC | Electrical conductivity | Potential soil salinity |
| Phosphorus | Available phosphorus status | Insufficient phosphorus |
| Potassium | Available potassium status | Insufficient potassium |
| Organic Matter | Organic matter status | Low organic matter |
| Soil Texture | Clay, sand, and silt characteristics | Drainage and water-retention limitations |
| SOC | Soil organic carbon | Low carbon content |
| Nitrogen | Modelled nitrogen content | Potential nitrogen limitation |
| CEC | Cation exchange capacity | Nutrient-retention limitations |
| Bulk Density | Soil density | Potential compaction |
| Water Content | Water retention at specified pressures | Water availability characteristics |

### 7.3 Assessment Rules

The system will use documented agricultural reference ranges to interpret soil properties.

Each property assessment should contain:

- Retrieved value or interval.
- Measurement unit, where confirmed.
- Data source.
- Applicable reference range.
- Assessment category.
- Explanation of the result.
- Data-quality or uncertainty warning.

Possible assessment categories:

1. Low
2. Moderate
3. High
4. Potential concern
5. Insufficient data

These categories are not universal thresholds. Their actual numerical boundaries must be selected according to verified units, laboratory methods, soil conditions, and appropriate agricultural references.

### 7.4 Multi-Property Soil Assessment

After assessing individual properties, the system will produce a combined soil-condition profile.

For example:

- Elevated EC may indicate a potential salinity problem.
- Unfavourable pH may limit the availability of certain nutrients.
- Low organic matter may indicate a need for improved organic inputs.
- Soil texture can influence drainage and water retention.
- Low CEC may indicate limited capacity to retain certain nutrients.

These indicators should be interpreted together rather than independently.

### 7.5 Risk Identification

The system will identify potential soil-related risks, including:

- Salinity risk
- Unfavourable soil pH
- Possible nutrient limitations
- Low organic matter
- Potential drainage limitations
- Potential compaction
- Water-retention limitations

A risk must only be reported when supported by relevant soil-property evidence.

For example, EC can support a salinity assessment, but it cannot independently establish the cause of salinity or determine the exact treatment required.

### 7.6 Soil Health Output

The assessment module will produce a structured result containing:

- Individual property assessments
- Potential soil limitations
- Identified risks
- Supporting evidence
- Data-source information
- Missing-data indicators
- Confidence or uncertainty information

A single overall soil-health score may be introduced later, but only after its indicators, weights, and validation methodology have been defined.

The system must not generate an arbitrary score that gives an impression of scientific accuracy.



## 8. Crop Suitability Assessment Methodology

### 8.1 Objective

The crop-suitability module will evaluate which crops are potentially suitable for a given geographical location based on available soil and environmental information.

The system will rank candidate crops and explain the factors influencing their suitability.

### 8.2 Input Features

The module will use the following information:

| Feature | Source | Purpose |
|---|---|---|
| Soil pH | APNA Farm + SoilGrids | Evaluate crop pH requirements |
| Electrical Conductivity | APNA Farm | Assess potential salinity limitations |
| Phosphorus | APNA Farm | Evaluate phosphorus availability |
| Potassium | APNA Farm | Evaluate potassium availability |
| Organic Matter | APNA Farm | Assess soil organic matter |
| Soil Texture | APNA Farm + SoilGrids | Evaluate soil physical suitability |
| Soil Organic Carbon | SoilGrids | Assess soil carbon characteristics |
| Nitrogen | SoilGrids | Provide additional nutrient information |
| CEC | SoilGrids | Evaluate nutrient-retention capacity |
| Temperature | Weather or climate data | Evaluate crop temperature requirements |
| Rainfall | Weather or climate data | Evaluate water requirements |
| Irrigation Availability | User input or external data | Assess water accessibility |

### 8.3 Crop Requirement Database

The system will require a structured database containing the agricultural requirements of candidate crops.

Each crop record should include:

- Crop name
- Suitable soil pH range
- Electrical conductivity tolerance
- Soil texture preferences
- Nutrient requirements
- Temperature requirements
- Rainfall requirements
- Water and irrigation requirements
- Growing-season information

Crop requirements must come from credible agricultural references relevant to Punjab and Pakistan wherever possible.

### 8.4 Suitability Evaluation

For each candidate crop:

1. Retrieve its documented agricultural requirements.
2. Compare those requirements with the available soil properties.
3. Evaluate environmental suitability using climate and water information.
4. Identify favourable and limiting factors.
5. Calculate a suitability score only when the scoring rules and required inputs are sufficiently supported.
6. Rank crops according to their resulting suitability.

### 8.5 Suitability Categories

The system may classify crops into:

- Highly Suitable
- Suitable
- Moderately Suitable
- Limited Suitability
- Insufficient Data

These categories will be based on documented crop requirements and a validated scoring method.

### 8.6 Explainable Recommendations

Every crop recommendation should explain:

- Why the crop is recommended.
- Which soil properties support its suitability.
- Which properties may limit its growth.
- Whether climate and irrigation conditions were considered.
- What additional information is needed.

For example, a crop may have favourable soil pH and texture but still receive a lower suitability assessment because rainfall or irrigation requirements cannot be satisfied.

### 8.7 Initial Development Strategy

The first version will implement soil-based crop screening using available APNA Farm and SoilGrids properties.

Climate and irrigation information will be integrated as additional inputs when reliable datasets become available.

The system must label soil-only recommendations as preliminary rather than presenting them as complete agricultural suitability predictions.

### 8.8 Final Output

For a given coordinate, the crop-suitability module will return:

- Ranked candidate crops
- Suitability category
- Supporting soil properties
- Potential limitations
- Missing environmental information
- Explanation for each recommendation



## 9. Fertilizer and Soil-Management Recommendation Methodology

### 9.1 Objective

The recommendation module will generate evidence-based fertilizer and soil-management guidance using the retrieved soil properties, identified limitations, and selected crop.

Its purpose is to help users understand what actions may improve soil conditions and support crop production.

### 9.2 Required Inputs

The module will use:

- Soil pH
- Electrical Conductivity (EC)
- Available phosphorus
- Available potassium
- Organic Matter
- Soil Texture
- Soil Organic Carbon (SOC)
- Nitrogen
- Cation Exchange Capacity (CEC)
- Selected crop
- Crop nutrient requirements
- Irrigation and water availability, when known

### 9.3 Nutrient Management

The system will assess available nutrient information and identify potential nutrient limitations.

| Nutrient or property | Possible management guidance |
|---|---|
| Nitrogen | Assess crop nitrogen requirements and available soil information |
| Phosphorus | Identify potential phosphorus deficiency |
| Potassium | Identify potential potassium deficiency |
| Organic Matter | Consider suitable organic amendments and residue management |
| SOC | Support long-term soil carbon management |

Fertilizer recommendations must account for crop requirements, confirmed nutrient units, soil-test methodology, and local agricultural guidelines.

### 9.4 Soil pH Management

The system will identify potentially acidic or alkaline soil conditions.

Possible management considerations include:

- Liming for appropriately diagnosed acidic soils.
- Evaluation of sodicity and soil amendments where relevant.
- Organic matter management.
- Selection of crops tolerant to the identified soil conditions.

A high pH value alone does not prove that gypsum is required. Sodicity must be evaluated using appropriate evidence, such as exchangeable sodium percentage or sodium adsorption ratio.

### 9.5 Salinity Management

Where EC indicates potential salinity:

- Identify the potential severity of the problem.
- Consider crop salt tolerance.
- Evaluate irrigation-water quality.
- Assess drainage conditions.
- Consider leaching only when water quality, drainage, and local conditions support it.

### 9.6 Soil Physical Management

Soil texture, bulk density, and water-retention information may support recommendations such as:

- Improving organic matter.
- Adjusting irrigation practices.
- Managing drainage.
- Reducing unnecessary soil compaction.
- Selecting crops appropriate for the soil's physical characteristics.

### 9.7 Recommendation Evidence

Each recommendation should contain:

- Identified soil condition.
- Supporting soil-property values.
- Data source.
- Recommended management action.
- Reason for the recommendation.
- Important limitations or additional tests required.

### 9.8 Fertilizer Quantity Calculation

Exact fertilizer quantities will be introduced only after the system has:

1. Confirmed nutrient measurement units and extraction methods.
2. Obtained reliable crop nutrient requirements.
3. Established local fertilizer recommendation guidelines.
4. Defined nutrient conversion and fertilizer-grade calculations.
5. Validated recommendations against agricultural expertise or field observations.

The system must not generate arbitrary fertilizer doses from unverified map estimates.

### 9.9 Final Output

The recommendation module will generate:

- Potential nutrient limitations.
- Suggested fertilizer-management considerations.
- Organic matter management guidance.
- Salinity and pH management considerations.
- Irrigation and drainage guidance.
- Crop-specific management advice.
- Evidence and uncertainty information.



## 10. Artificial Intelligence and Machine Learning Methodology

### 10.1 Objective

The AI/ML component will support soil-property analysis, soil-condition assessment, risk identification, and crop-suitability prediction.

The model-development strategy will depend on the availability, quality, and reliability of training data.

### 10.2 Available Data for Model Development

The initial system has two major data sources:

**APNA Farm**
- Digitized pH classes.
- Digitized electrical conductivity classes.
- Digitized phosphorus classes.
- Digitized potassium classes.
- Digitized organic matter classes.
- Digitized soil texture classes.

**SoilGrids**
- Modelled soil-property estimates.
- Multiple soil-depth intervals.
- Spatially continuous coverage across Punjab.
- Soil physical, chemical, and carbon-related properties.

These datasets provide useful features for spatial analysis and exploratory modelling. However, their estimates must not automatically be treated as verified laboratory targets.

### 10.3 Potential Machine Learning Tasks

| Task | Possible approach | Objective |
|---|---|---|
| Soil-property estimation | Regression | Estimate continuous soil properties |
| Soil classification | Classification | Categorize soil conditions |
| Soil-risk identification | Rule-based + classification | Identify potential soil limitations |
| Crop suitability | Ranking or classification | Rank crops by suitability |
| Spatial pattern analysis | Clustering | Identify areas with similar soil characteristics |

The final selection will depend on the availability of reliable target labels.

### 10.4 Initial Model Candidates

The following models may be evaluated where the data supports supervised learning:

**Traditional Machine Learning**
- Random Forest
- XGBoost
- Gradient Boosting
- Decision Tree
- Support Vector Machine

**Deep Learning**
- Multilayer Perceptron (MLP)
- Convolutional Neural Network (CNN)
- Spatial or geospatial deep learning architectures, if sufficient spatial training data becomes available.

### 10.5 Model Selection Strategy

The system will not select a model based only on popularity or complexity.

Model selection will consider:

- Dataset size and quality.
- Availability of reliable target variables.
- Spatial distribution of training examples.
- Prediction requirements.
- Computational resources.
- Model interpretability.
- Generalization to unseen locations.

Traditional machine learning will be evaluated before more complex deep learning architectures.

### 10.6 Avoiding Data Leakage

Spatial data requires particular attention during model validation.

Randomly splitting neighbouring map cells between training and testing can produce overly optimistic results because nearby locations may have similar soil characteristics.

Where sufficient independent data exists, spatially separated training, validation, and testing regions will be preferred.

### 10.7 Ground-Truth Validation

Model predictions should eventually be evaluated against independent soil laboratory measurements or verified field observations.

Potential validation sources include:

- Historical soil-testing laboratory records.
- New soil samples collected in Sahiwal.
- Laboratory measurements of pH, EC, phosphorus, potassium, and organic matter.
- Verified field observations.

SoilGrids estimates must not be used as independent laboratory ground truth when the objective is to validate actual soil conditions.

### 10.8 Model Evaluation

Evaluation metrics will depend on the task.

**Regression**
- MAE
- RMSE
- R²

**Classification**
- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

**Crop Suitability Ranking**
- Ranking agreement with agricultural expert assessments.
- Precision at K, where suitable reference rankings are available.

### 10.9 Initial Development Decision

The first SoilSenseAI version will prioritize a transparent, rule-based soil assessment and crop-screening engine.

Machine learning models will be introduced for specific prediction tasks when appropriate training labels and validation data are available.

Deep learning will be considered when dataset size, spatial structure, and target quality justify its use.

The goal is reliable agricultural intelligence rather than unnecessary model complexity.



## 11. Validation and Field-Testing Methodology

### 11.1 Objective

The validation process will evaluate the reliability of SoilSenseAI's soil assessments, spatial predictions, crop recommendations, and management guidance.

The system will use both computational validation and independent field or laboratory observations.

### 11.2 Validation Strategy

Validation will take place at three levels:

**Level 1 — Data Validation**
- Check coordinate validity and geographic coverage.
- Identify missing and anomalous values.
- Verify measurement units and source metadata.
- Check spatial alignment between datasets.
- Preserve source-specific uncertainty.

**Level 2 — Computational Validation**
- Evaluate data-processing operations.
- Test coordinate-based soil-property retrieval.
- Verify soil assessment rules against documented agricultural reference ranges.
- Evaluate machine learning models using appropriate validation metrics.
- Test the system using locations not used during model development.

**Level 3 — Field and Laboratory Validation**
- Select representative locations in Sahiwal.
- Retrieve SoilSenseAI predictions for those coordinates.
- Collect soil samples from corresponding locations.
- Obtain laboratory measurements.
- Compare predicted or estimated properties with laboratory results.
- Document differences and possible sources of error.

### 11.3 Sahiwal Pilot

Sahiwal will be used as the initial field-validation district.

The pilot will help determine whether the system's coordinate-based assessments reflect actual local soil conditions.

Potential sampling locations should represent different soil conditions and spatial areas.

For each sampling location, record:

- Sample identifier
- Latitude and longitude
- Collection date
- Sampling depth
- Soil sampling method
- Laboratory test results
- Relevant field observations

### 11.4 Laboratory Comparison

Where laboratory measurements are available, compare them with corresponding SoilSenseAI estimates.

Potential comparison properties include:

- Soil pH
- Electrical Conductivity
- Available Phosphorus
- Available Potassium
- Organic Matter

The comparison must account for differences in sampling depth, laboratory extraction methods, measurement units, and spatial resolution.

APNA Farm and SoilGrids comparisons alone do not establish laboratory accuracy.

### 11.5 Evaluation Metrics

For continuous soil-property estimates:

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- R²

For classification tasks:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

For crop recommendations:

- Agreement with documented crop requirements.
- Expert assessment of suitability rankings.
- Field observations where available.

### 11.6 Spatial Validation

Because nearby locations can have similar soil characteristics, spatial dependence must be considered.

Where sufficient independent observations exist:

- Separate training and testing locations geographically.
- Avoid placing neighbouring samples in both training and testing sets.
- Evaluate performance on previously unseen areas.
- Report performance separately for different geographical regions where possible.

### 11.7 Continuous Improvement

Validation findings will be used to:

- Correct data-processing errors.
- Improve spatial matching.
- Refine soil assessment thresholds.
- Improve crop suitability rules.
- Retrain models when reliable new training data becomes available.
- Update recommendation logic.

### 11.8 Validation Principle

SoilSenseAI must distinguish between:

1. A spatial estimate obtained from an existing dataset.
2. A prediction generated by an AI/ML model.
3. An independently verified laboratory measurement.

The system must report these categories clearly and must not claim field accuracy without independent validation evidence.



## 12. Overall System Architecture

### 12.1 Architecture Overview

SoilSenseAI will use a modular architecture consisting of independent components responsible for data retrieval, spatial processing, soil assessment, prediction, recommendations, and visualization.

The modules will communicate through structured data rather than duplicating processing logic.

### 12.2 Main System Components

**1. User Interface**
- Accepts latitude and longitude.
- Displays soil assessments and recommendations.
- Presents maps and spatial information.
- Communicates missing data and uncertainty.

**2. Coordinate Processing Module**
- Validates geographic coordinates.
- Checks Punjab administrative boundaries.
- Performs spatial matching.
- Identifies the relevant soil-data locations.

**3. Data Integration Module**
- Retrieves APNA Farm properties.
- Retrieves SoilGrids properties and depth intervals.
- Maintains source information.
- Handles missing values and unit conversions.

**4. Soil Intelligence Engine**
- Processes retrieved soil properties.
- Evaluates individual soil conditions.
- Identifies potential soil risks.
- Produces a structured soil profile.

**5. AI/ML Prediction Layer**
- Hosts trained models when suitable training data becomes available.
- Performs supported prediction tasks.
- Returns predictions with evaluation and uncertainty information.

**6. Crop Suitability Module**
- Compares soil conditions with crop requirements.
- Integrates climate and irrigation information when available.
- Produces ranked crop recommendations.

**7. Recommendation Engine**
- Generates soil-management guidance.
- Identifies potential nutrient limitations.
- Provides crop-specific recommendations.
- Avoids unsupported fertilizer quantities.

**8. Visualization and Reporting Module**
- Displays soil-property maps.
- Presents soil assessment results.
- Shows crop rankings and potential risks.
- Reports data sources and limitations.

### 12.3 Data Flow

The system follows this sequence:

1. User submits coordinates.
2. Coordinate processing validates the location.
3. Data integration retrieves available soil information.
4. Soil Intelligence Engine evaluates the soil profile.
5. AI/ML models perform supported prediction tasks.
6. Crop suitability evaluates candidate crops.
7. Recommendation Engine generates management guidance.
8. Visualization module presents the final assessment.

### 12.4 Modular Development Principle

Each component will be developed and tested independently.

This approach allows individual modules to be improved without redesigning the entire application.

### 12.5 Future Expansion

The architecture should support future integration of:

- Weather and climate datasets.
- Irrigation availability.
- Soil laboratory records.
- Satellite imagery.
- Field observations.
- Additional machine learning models.
- New agricultural datasets.



## 13. Implementation Roadmap

SoilSenseAI will be developed incrementally, prioritizing a functional coordinate-based application.

### Phase 1 — Data Foundation
- Organize APNA Farm and SoilGrids datasets.
- Standardize coordinate reference systems.
- Verify units and metadata.
- Implement coordinate-based data retrieval.
- Handle missing values and data quality.

### Phase 2 — Soil Intelligence Engine
- Implement individual soil-property assessments.
- Define evidence-based soil condition categories.
- Identify potential soil risks.
- Generate structured soil profiles.

### Phase 3 — Crop Suitability
- Build a crop requirement database.
- Implement soil-based crop screening.
- Add climate and irrigation information.
- Generate explainable crop rankings.

### Phase 4 — Recommendations
- Implement soil-management guidance.
- Integrate crop-specific nutrient requirements.
- Add validated fertilizer calculations when sufficient data is available.

### Phase 5 — AI/ML Development
- Identify suitable prediction tasks and target labels.
- Establish training and validation datasets.
- Evaluate traditional machine learning models.
- Develop deep learning models only when justified.

### Phase 6 — Application Development
- Build a user interface.
- Integrate maps and coordinate input.
- Connect all intelligence modules.
- Display assessments, risks, and recommendations.

### Phase 7 — Validation and Deployment
- Conduct Sahiwal pilot validation.
- Compare estimates with independent laboratory observations.
- Improve system accuracy and reliability.
- Prepare the Punjab-wide application.
